```python

# TASK 4: FINITE STATE AUTOMATA

# ---------------------------------
# 1. DFA - Strings ending with "ab"
# ---------------------------------

print("1. DFA IMPLEMENTATION")
print("---------------------")

dfa_transitions = {
    "q0": {"a": "q1", "b": "q0"},
    "q1": {"a": "q1", "b": "q2"},
    "q2": {"a": "q1", "b": "q0"}
}

def dfa_accept(word):
    current_state = "q0"

    for symbol in word:
        if symbol not in ["a", "b"]:
            return False

        current_state = dfa_transitions[current_state][symbol]

    return current_state == "q2"


test_words = ["ab", "aab", "bab", "abb", "aaa"]

for word in test_words:
    result = "Accepted" if dfa_accept(word) else "Rejected"
    print(word, ":", result)


# ---------------------------------
# 2. NFA - Strings containing "ab"
# ---------------------------------

print("\n2. NFA IMPLEMENTATION")
print("---------------------")

nfa_transitions = {
    "q0": {
        "a": {"q0", "q1"},
        "b": {"q0"}
    },
    "q1": {
        "a": set(),
        "b": {"q2"}
    },
    "q2": {
        "a": {"q2"},
        "b": {"q2"}
    }
}

def nfa_accept(word):
    current_states = {"q0"}

    for symbol in word:
        next_states = set()

        for state in current_states:
            next_states.update(
                nfa_transitions[state][symbol]
            )

        current_states = next_states

    return "q2" in current_states


test_words = ["ab", "aab", "bba", "baab", "bbb"]

for word in test_words:
    result = "Accepted" if nfa_accept(word) else "Rejected"
    print(word, ":", result)


# ---------------------------------
# 3. NFA TO DFA CONVERSION
#    Subset Construction Method
# ---------------------------------

print("\n3. NFA TO DFA CONVERSION")
print("------------------------")

alphabet = ["a", "b"]
start_state = frozenset({"q0"})

dfa_states = [start_state]
dfa_transitions = {}

index = 0

while index < len(dfa_states):

    current_set = dfa_states[index]
    dfa_transitions[current_set] = {}

    for symbol in alphabet:

        next_states = set()

        for state in current_set:
            next_states.update(
                nfa_transitions[state][symbol]
            )

        next_set = frozenset(next_states)

        dfa_transitions[current_set][symbol] = next_set

        if next_set not in dfa_states:
            dfa_states.append(next_set)

    index += 1


print("DFA States:")

for state in dfa_states:
    print(set(state))


print("\nDFA Transition Table:")

for state, moves in dfa_transitions.items():

    for symbol, destination in moves.items():

        print(
            set(state),
            "--", symbol, "-->",
            set(destination)
        )


# Identify final states of converted DFA

final_states = [
    state for state in dfa_states
    if "q2" in state
]

print("\nFinal States of DFA:")

for state in final_states:
    print(set(state))


# ---------------------------------
# 4. VALIDATE USING CONVERTED DFA
# ---------------------------------

print("\n4. VALIDATION USING CONVERTED DFA")
print("--------------------------------")

def converted_dfa_accept(word):

    current_state = start_state

    for symbol in word:

        if symbol not in alphabet:
            return False

        current_state = dfa_transitions[current_state][symbol]

    return "q2" in current_state


test_words = ["ab", "aab", "bba", "baab", "bbb"]

for word in test_words:

    result = (
        "Accepted"
        if converted_dfa_accept(word)
        else "Rejected"
    )

    print(word, ":", result)
```

```python

# TASK 7: FINITE-STATE MORPHOLOGICAL PARSING

# 1. LEXICON

print("1. MORPHOLOGICAL LEXICON")
print("-----------------------")

# Each stem is assigned a grammatical category.

lexicon = {
    "patient": ["NOUN"],
    "doctor": ["NOUN"],
    "medicine": ["NOUN"],
    "record": ["NOUN", "VERB"],
    "treat": ["VERB"],
    "monitor": ["VERB"]
}

for stem, categories in lexicon.items():
    print(stem, ":", ", ".join(categories))


# -----------------------------------------
# 2. CONTINUATION CLASSES
# -----------------------------------------

print("\n2. CONTINUATION CLASSES")
print("-----------------------")

continuation_classes = {
    "NOUN": {
        "s": "PLURAL"
    },

    "VERB": {
        "s": "PRESENT_3SG",
        "ed": "PAST",
        "ing": "PRESENT_PARTICIPLE"
    }
}

for category, suffixes in continuation_classes.items():

    print(category, "->", suffixes)


# -----------------------------------------
# 3. FINITE-STATE TRANSDUCER
# -----------------------------------------

print("\n3. FINITE-STATE TRANSDUCER")
print("--------------------------")

def morphological_analysis(word):

    analyses = []

    for stem, categories in lexicon.items():

        # Check whether the word begins with the stem.
        if not word.startswith(stem):
            continue

        suffix = word[len(stem):]

        for category in categories:

            # Base form
            if suffix == "":
                analyses.append(
                    stem + "+" + category
                )

            # Inflected forms
            elif suffix in continuation_classes[category]:

                grammatical_feature = (
                    continuation_classes[category][suffix]
                )

                analyses.append(
                    stem + "+" + category +
                    "+" + grammatical_feature
                )

    return analyses


# -----------------------------------------
# 4. ANALYSE WORD FORMS
# -----------------------------------------

print("\n4. MORPHOLOGICAL ANALYSIS")
print("------------------------")

test_words = [
    "patients",
    "doctors",
    "medicines",
    "treated",
    "treating",
    "records",
    "monitoring"
]

for word in test_words:

    result = morphological_analysis(word)

    print("\nWord:", word)

    if result:
        for analysis in result:
            print("Analysis:", analysis)
    else:
        print("No analysis found")


# -----------------------------------------
# 5. MORPHOLOGICAL GENERATION
# -----------------------------------------

print("\n5. MORPHOLOGICAL GENERATION")
print("---------------------------")

def generate_word(stem, category, feature):

    if stem not in lexicon:
        return "Stem not found"

    if category not in lexicon[stem]:
        return "Invalid grammatical category"

    if feature == "BASE":
        return stem

    for suffix, grammatical_feature in (
        continuation_classes[category].items()
    ):

        if grammatical_feature == feature:
            return stem + suffix

    return "Invalid grammatical feature"


generation_examples = [
    ("patient", "NOUN", "PLURAL"),
    ("doctor", "NOUN", "PLURAL"),
    ("treat", "VERB", "PAST"),
    ("treat", "VERB", "PRESENT_PARTICIPLE"),
    ("record", "VERB", "PRESENT_3SG")
]

for stem, category, feature in generation_examples:

    generated = generate_word(
        stem, category, feature
    )

    print(
        stem + "+" + category + "+" + feature,
        "->",
        generated
    )


# -----------------------------------------
# 6. MORPHOLOGICAL AMBIGUITY
# -----------------------------------------

print("\n6. MORPHOLOGICAL AMBIGUITY")
print("--------------------------")

ambiguous_word = "records"

analyses = morphological_analysis(ambiguous_word)

print("Word:", ambiguous_word)

for analysis in analyses:
    print("Possible Analysis:", analysis)

if len(analyses) > 1:
    print("Result: Morphologically Ambiguous")
else:
    print("Result: Unambiguous")


# -----------------------------------------
# 7. FINAL SUMMARY
# -----------------------------------------

print("\n7. FINAL SUMMARY")
print("----------------")

print("Lexicon created successfully.")
print("Continuation classes defined.")
print("Finite-State Transducer implemented.")
print("Word forms analysed successfully.")
print("New word forms generated successfully.")
print("Morphological ambiguity identified.")
```

```python

# TASK 8: ORTHOGRAPHIC RULES AND TWO-LEVEL MORPHOLOGY

# -----------------------------------------
# 1. ORTHOGRAPHIC REWRITE RULES
# -----------------------------------------

print("1. ORTHOGRAPHIC REWRITE RULES")
print("-----------------------------")

# Rule 1: Change consonant + y into ies
def plural_y(word):
    if word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "ies"
    return word + "s"


# Rule 2: Add es to words ending in s, x, ch or sh
def plural_es(word):
    if word.endswith(("s", "x", "ch", "sh")):
        return word + "es"
    return word + "s"


# Rule 3: Remove final e before adding ing
def remove_e(word):
    if word.endswith("e") and not word.endswith("ee"):
        return word[:-1] + "ing"
    return word + "ing"


# Rule 4: Change ie into ying
def ie_to_ying(word):
    if word.endswith("ie"):
        return word[:-2] + "ying"
    return word + "ing"


# Rule 5: Double final consonant before adding ing
def double_consonant(word):
    vowels = "aeiou"

    if (len(word) >= 3 and
        word[-1] not in vowels and
        word[-1] not in "wxy" and
        word[-2] in vowels and
        word[-3] not in vowels):

        return word + word[-1] + "ing"

    return word + "ing"


print("Rule 1 - city ->", plural_y("city"))
print("Rule 2 - watch ->", plural_es("watch"))
print("Rule 3 - make ->", remove_e("make"))
print("Rule 4 - lie ->", ie_to_ying("lie"))
print("Rule 5 - stop ->", double_consonant("stop"))


# -----------------------------------------
# 2. FINITE-STATE TRANSDUCER
# -----------------------------------------

print("\n2. FINITE-STATE TRANSDUCER")
print("--------------------------")

# A simple transducer maps a lexical form
# and a grammatical feature to a surface form.

def fst_generate(stem, feature):

    if feature == "PLURAL":

        if stem.endswith("y") and stem[-2] not in "aeiou":
            return plural_y(stem)

        elif stem.endswith(("s", "x", "ch", "sh")):
            return plural_es(stem)

        else:
            return stem + "s"

    elif feature == "PAST":
        if stem.endswith("e"):
            return stem + "d"
        elif stem.endswith("y") and stem[-2] not in "aeiou":
            return stem[:-1] + "ied"
        else:
            return stem + "ed"

    elif feature == "ING":

        if stem.endswith("ie"):
            return ie_to_ying(stem)

        elif stem.endswith("e") and not stem.endswith("ee"):
            return remove_e(stem)

        elif (len(stem) >= 3 and
              stem[-1] not in "aeiouwxy" and
              stem[-2] in "aeiou" and
              stem[-3] not in "aeiou"):
            return double_consonant(stem)

        else:
            return stem + "ing"

    return "Invalid feature"


examples = [
    ("city", "PLURAL"),
    ("watch", "PLURAL"),
    ("make", "ING"),
    ("lie", "ING"),
    ("stop", "ING"),
    ("play", "PAST")
]

for stem, feature in examples:

    result = fst_generate(stem, feature)

    print(stem + " + " + feature, "->", result)


# -----------------------------------------
# 3. CASCADING OF ORTHOGRAPHIC RULES
# -----------------------------------------

print("\n3. CASCADING OF RULES")
print("---------------------")

# Rules are applied in sequence.
# First, identify the grammatical feature.
# Then, apply the corresponding spelling rule.

def apply_cascade(stem, feature):

    print("Input:", stem, "+", feature)

    if feature == "ING":

        if stem.endswith("ie"):
            output = ie_to_ying(stem)

        elif stem.endswith("e") and not stem.endswith("ee"):
            output = remove_e(stem)

        elif (len(stem) >= 3 and
              stem[-1] not in "aeiouwxy" and
              stem[-2] in "aeiou" and
              stem[-3] not in "aeiou"):
            output = double_consonant(stem)

        else:
            output = stem + "ing"

    elif feature == "PLURAL":

        if stem.endswith("y") and stem[-2] not in "aeiou":
            output = plural_y(stem)

        elif stem.endswith(("s", "x", "ch", "sh")):
            output = plural_es(stem)

        else:
            output = stem + "s"

    else:
        output = "Unsupported feature"

    print("After rule application:", output)
    return output


apply_cascade("study", "ING")
apply_cascade("city", "PLURAL")
apply_cascade("stop", "ING")


# -----------------------------------------
# 4. TWO-LEVEL MORPHOLOGY
# -----------------------------------------

print("\n4. TWO-LEVEL MORPHOLOGY")
print("-----------------------")

# Two-level morphology represents the relationship
# between lexical symbols and surface symbols.

two_level_examples = {
    "city + s": "cities",
    "make + ing": "making",
    "lie + ing": "lying",
    "stop + ing": "stopping",
    "watch + s": "watches"
}

for lexical, surface in two_level_examples.items():

    print("Lexical form:", lexical)
    print("Surface form:", surface)
    print()


# -----------------------------------------
# 5. LEXICAL AND SURFACE FORM COMPARISON
# -----------------------------------------

print("5. LEXICAL-SURFACE FORM MAPPING")
print("--------------------------------")

mappings = [
    ("city+s", "cities"),
    ("make+ing", "making"),
    ("lie+ing", "lying"),
    ("stop+ing", "stopping"),
    ("watch+s", "watches")
]

for lexical, surface in mappings:

    print(lexical, "---->", surface)


# -----------------------------------------
# 6. COMPARISON OF MORPHOLOGICAL MODELS
# -----------------------------------------

print("\n6. MODEL COMPARISON")
print("-------------------")

print("Rule-Based Morphology:")
print("- Applies spelling rules sequentially.")
print("- Easy to understand and implement.")

print("\nTwo-Level Morphology:")
print("- Represents lexical and surface forms together.")
print("- Uses constraints between symbol pairs.")
print("- Multiple spelling changes can be represented.")

print("\n7. FINAL SUMMARY")
print("----------------")

print("Five orthographic rules demonstrated.")
print("Finite-State Transducer implemented.")
print("Cascading of spelling rules demonstrated.")
print("Lexical and surface forms mapped.")
print("Two-level morphology explained.")
```

```python

# TASK 9: PORTER STEMMER

import nltk
from nltk.stem import PorterStemmer

# Create Porter Stemmer object
stemmer = PorterStemmer()

# -----------------------------------------
# 1. PORTER STEMMING
# -----------------------------------------

print("1. PORTER STEMMING")
print("------------------")

words = [
    "caresses",
    "ponies",
    "cats",
    "running",
    "hopping",
    "happy",
    "relational",
    "conditional",
    "digitizer",
    "activation"
]

for word in words:
    print(word, "->", stemmer.stem(word))


# -----------------------------------------
# 2. PORTER STEMMER CONDITIONS
# -----------------------------------------

print("\n2. PORTER STEMMER CONDITIONS")
print("----------------------------")

# Convert a word into consonant/vowel representation.
# y is treated as a consonant in this simplified
# representation.

def cv_sequence(word):

    result = ""

    for i, char in enumerate(word):

        if char in "aeiou":
            result += "V"

        elif char == "y" and i > 0 and word[i - 1] in "aeiou":
            result += "V"

        else:
            result += "C"

    return result


# Calculate measure m:
# Number of VC sequences in the word.

def calculate_m(word):

    sequence = cv_sequence(word)

    count = 0

    for i in range(len(sequence) - 1):

        if sequence[i:i + 2] == "VC":
            count += 1

    return count


# *v: stem contains a vowel

def contains_vowel(word):
    return any(char in "aeiou" for char in word)


# *d: stem ends with a double consonant

def double_consonant(word):

    return (
        len(word) >= 2 and
        word[-1] == word[-2] and
        word[-1] not in "aeiou"
    )


# *o: stem ends in consonant-vowel-consonant,
# excluding w, x and y as the final consonant.

def cvc_condition(word):

    if len(word) < 3:
        return False

    sequence = cv_sequence(word)

    return (
        sequence[-3:] == "CVC" and
        word[-1] not in "wxy"
    )


test_words = ["tr", "trouble", "tree", "running", "hop"]

for word in test_words:

    print("\nWord:", word)
    print("CV Pattern:", cv_sequence(word))
    print("Measure m:", calculate_m(word))
    print("Contains vowel (*v):", contains_vowel(word))
    print("Double consonant (*d):", double_consonant(word))
    print("CVC condition (*o):", cvc_condition(word))


# -----------------------------------------
# 3. FIVE MAJOR PORTER STEMMING STEPS
# -----------------------------------------

print("\n3. FIVE MAJOR PORTER STEMMING STEPS")
print("-----------------------------------")

# Examples illustrating the five stages.
# These transformations are shown separately
# to explain the Porter algorithm.

step_examples = {
    "Step 1 - Plural and tense handling": [
        ("caresses", "caress"),
        ("ponies", "poni"),
        ("cats", "cat"),
        ("running", "run"),
        ("hopping", "hop")
    ],

    "Step 2 - Derivational suffix replacement": [
        ("relational", "relate"),
        ("conditional", "condition"),
        ("digitizer", "digitize")
    ],

    "Step 3 - Further suffix reduction": [
        ("triplicate", "triplic"),
        ("formative", "form"),
        ("formalize", "formal")
    ],

    "Step 4 - Suffix removal": [
        ("revival", "reviv"),
        ("allowance", "allow"),
        ("adjustment", "adjust")
    ],

    "Step 5 - Final e and l removal": [
        ("probate", "probat"),
        ("rate", "rate"),
        ("controll", "control")
    ]
}

for step, examples in step_examples.items():

    print("\n" + step)

    for original, result in examples:
        print(original, "->", result)


# -----------------------------------------
# 4. STEMMING VS LEMMATISATION
# -----------------------------------------

print("\n4. STEMMING VS LEMMATISATION")
print("----------------------------")

# Simple dictionary-based lemmatisation
# for demonstration.

lemma_dictionary = {
    "running": "run",
    "ran": "run",
    "better": "good",
    "mice": "mouse",
    "children": "child",
    "studies": "study",
    "cars": "car"
}

comparison_words = [
    "running",
    "ran",
    "better",
    "mice",
    "children",
    "studies",
    "cars"
]

for word in comparison_words:

    stem = stemmer.stem(word)

    lemma = lemma_dictionary.get(word, word)

    print("\nWord:", word)
    print("Stem:", stem)
    print("Lemma:", lemma)


# -----------------------------------------
# 5. STEMMING ERRORS
# -----------------------------------------

print("\n5. STEMMING ERRORS")
print("------------------")

error_examples = [
    ("university", "univers", "Over-stemming"),
    ("organization", "organ", "Possible over-stemming"),
    ("policy", "polici", "Under-stemming"),
    ("running", "run", "Correct stemming")
]

for word, stem, error_type in error_examples:

    print(word, "->", stem, ":", error_type)


# -----------------------------------------
# 6. INFORMATION RETRIEVAL EVALUATION
# -----------------------------------------

print("\n6. INFORMATION RETRIEVAL EVALUATION")
print("-----------------------------------")

# Documents retrieved by the search system
retrieved_documents = {"D1", "D2", "D3", "D4"}

# Documents actually relevant to the query
relevant_documents = {"D1", "D2", "D5"}

# Calculate True Positives, False Positives
# and False Negatives.

true_positive = len(
    retrieved_documents.intersection(relevant_documents)
)

false_positive = len(
    retrieved_documents - relevant_documents
)

false_negative = len(
    relevant_documents - retrieved_documents
)

precision = true_positive / (
    true_positive + false_positive
)

recall = true_positive / (
    true_positive + false_negative
)

f1_score = (
    2 * precision * recall /
    (precision + recall)
)

print("True Positives:", true_positive)
print("False Positives:", false_positive)
print("False Negatives:", false_negative)

print("Precision:", round(precision, 2))
print("Recall:", round(recall, 2))
print("F1-Score:", round(f1_score, 2))


# -----------------------------------------
# 7. FINAL SUMMARY
# -----------------------------------------

print("\n7. FINAL SUMMARY")
print("----------------")

print("Porter Stemmer implemented successfully.")
print("Five major stemming stages demonstrated.")
print("Morphological conditions calculated.")
print("Stemming and lemmatisation compared.")
print("Stemming errors demonstrated.")
print("Information retrieval metrics calculated.")
```

```python

# TASK 10: IMPLEMENTATION AND TESTING OF NLP PIPELINE

import re
import time
import matplotlib.pyplot as plt
from nltk.stem import PorterStemmer

# Create Porter Stemmer
stemmer = PorterStemmer()


# -----------------------------------------
# 1. SAMPLE HOSPITAL RECORDS
# -----------------------------------------

print("1. SAMPLE HOSPITAL RECORDS")
print("--------------------------")

records = [
    "Pt has fever and SOB on 12/11/2026.",
    "BP 120/80 mmHg and temperature 38 C.",
    "Give 500 mg medicine to the patient.",
    "Dr reviewed the patient on 13/11/2026.",
    "Patient has DM and HTN.",
    "The patient received 250 mg medicine."
]

for i, record in enumerate(records, start=1):
    print("Record", i, ":", record)


# -----------------------------------------
# 2. ABBREVIATION EXPANSION
# -----------------------------------------

print("\n2. ABBREVIATION EXPANSION")
print("-------------------------")

abbreviations = {
    "pt": "patient",
    "dr": "doctor",
    "sob": "shortness of breath",
    "bp": "blood pressure",
    "dm": "diabetes mellitus",
    "htn": "hypertension"
}

def expand_abbreviations(text):

    for short_form, full_form in abbreviations.items():

        pattern = r"\b" + re.escape(short_form) + r"\b"

        text = re.sub(
            pattern,
            full_form,
            text,
            flags=re.IGNORECASE
        )

    return text


# -----------------------------------------
# 3. DATE NORMALISATION
# -----------------------------------------

print("\n3. DATE NORMALISATION")
print("--------------------")

def normalize_dates(text):

    # Assumes input date format is DD/MM/YYYY.

    def replace_date(match):

        day = match.group(1)
        month = match.group(2)
        year = match.group(3)

        return year + "-" + month.zfill(2) + "-" + day.zfill(2)

    return re.sub(
        r"\b(\d{1,2})/(\d{1,2})/(\d{4})\b",
        replace_date,
        text
    )


# -----------------------------------------
# 4. MEDICAL UNIT NORMALISATION
# -----------------------------------------

print("\n4. MEDICAL UNIT NORMALISATION")
print("-----------------------------")

def normalize_units(text):

    return re.sub(
        r"(\d+(?:\.\d+)?)\s*(mg|g|ml|mmhg)\b",
        lambda match: match.group(1) + match.group(2),
        text,
        flags=re.IGNORECASE
    )


# -----------------------------------------
# 5. TOKENISATION
# -----------------------------------------

print("\n5. TOKENISATION")
print("---------------")

def tokenize(text):

    pattern = (
        r"\d{4}-\d{2}-\d{2}|"
        r"\d{2,3}/\d{2,3}(?:mmhg)?|"
        r"\d+(?:\.\d+)?(?:mg|ml|mmhg|g)?|"
        r"[a-z]+(?:-[a-z]+)*"
    )

    return re.findall(pattern, text.lower())


# -----------------------------------------
# 6. COMPLETE NLP PIPELINE
# -----------------------------------------

print("\n6. COMPLETE NLP PIPELINE")
print("------------------------")

def nlp_pipeline(text):

    # Convert text into lowercase
    text = text.lower()

    # Expand medical abbreviations
    text = expand_abbreviations(text)

    # Standardise dates
    text = normalize_dates(text)

    # Standardise medical units
    text = normalize_units(text)

    # Tokenise text
    tokens = tokenize(text)

    # Apply Porter stemming
    stems = [stemmer.stem(word) for word in tokens]

    return text, tokens, stems


# Process all records and measure execution time

start_time = time.perf_counter()

processed_records = []

for record in records:

    normalized, tokens, stems = nlp_pipeline(record)

    processed_records.append({
        "normalized": normalized,
        "tokens": tokens,
        "stems": stems
    })

end_time = time.perf_counter()

execution_time = end_time - start_time


# Display first processed record

print("Original Record:")
print(records[0])

print("\nNormalized Record:")
print(processed_records[0]["normalized"])

print("\nTokens:")
print(processed_records[0]["tokens"])

print("\nStemmed Words:")
print(processed_records[0]["stems"])


# -----------------------------------------
# 7. TEST CASE EVALUATION
# -----------------------------------------

print("\n7. TEST CASE EVALUATION")
print("-----------------------")

test_cases = [
    ("Pt has fever.", "patient has fever."),
    ("BP 120/80 mmHg", "blood pressure 120/80mmhg"),
    ("Review on 12/11/2026", "review on 2026-11-12"),
    ("Give 500 mg", "give 500mg"),
    ("Dr treated the patient.", "doctor treated the patient.")
]

passed = 0

for input_text, expected in test_cases:

    actual, tokens, stems = nlp_pipeline(input_text)

    if actual == expected:
        result = "PASS"
        passed += 1
    else:
        result = "FAIL"

    print("\nInput:", input_text)
    print("Expected:", expected)
    print("Actual:", actual)
    print("Result:", result)


total_tests = len(test_cases)
failed = total_tests - passed

accuracy = (passed / total_tests) * 100

print("\nTotal Test Cases:", total_tests)
print("Passed:", passed)
print("Failed:", failed)
print("Test Accuracy:", round(accuracy, 2), "%")


# -----------------------------------------
# 8. EXECUTION TIME AND PERFORMANCE
# -----------------------------------------

print("\n8. EXECUTION PERFORMANCE")
print("-------------------------")

total_tokens = sum(
    len(record["tokens"])
    for record in processed_records
)

print("Records Processed:", len(processed_records))
print("Total Tokens:", total_tokens)
print("Execution Time:", round(execution_time, 6), "seconds")


# -----------------------------------------
# 9. PERFORMANCE VISUALISATION
# -----------------------------------------

print("\n9. PERFORMANCE VISUALISATION")
print("-----------------------------")

labels = ["Passed Tests", "Failed Tests"]
values = [passed, failed]

plt.figure(figsize=(7, 5))

plt.bar(labels, values)

plt.title("NLP Pipeline Test Results")
plt.xlabel("Test Cases")
plt.ylabel("Number of Tests")

plt.show()


# -----------------------------------------
# 10. FINAL SUMMARY
# -----------------------------------------

print("\n10. FINAL SUMMARY")
print("-----------------")

print("Abbreviation expansion completed.")
print("Date normalisation completed.")
print("Medical unit normalisation completed.")
print("Tokenisation completed.")
print("Porter stemming completed.")
print("Test cases evaluated.")
print("Execution time measured.")
print("Performance graph generated.")
```

```python
#Task 11 – Compare and Recommend a Normalisation Design
from nltk.stem import PorterStemmer
import re

# Create Porter Stemmer
stemmer = PorterStemmer()

# Sample hospital records
documents = {
    "D1": "Patient treated with medicine",
    "D2": "Patients are treated",
    "D3": "Patient treatment record",
    "D4": "Doctor treated patient",
    "D5": "Medicines prescribed to patient"
}

# Search query
query = "patients treated"

# Manually identified relevant documents
relevant_docs = {"D1", "D2", "D4"}


# Tokenisation
def tokenize(text):
    return re.findall(r"[a-z]+", text.lower())


# Method 1: Exact word matching
def exact_normalize(word):
    return word


# Method 2: Porter stemming
def porter_normalize(word):
    return stemmer.stem(word)


# Method 3: Morphology-aware normalisation
lemma_dict = {
    "patients": "patient",
    "patient": "patient",
    "treated": "treat",
    "treating": "treat",
    "treats": "treat",
    "medicines": "medicine"
}

def morphology_normalize(word):
    return lemma_dict.get(word, word)


# Search function
def search_documents(normalizer):
    query_words = {
        normalizer(word) for word in tokenize(query)
    }

    results = []

    for doc_id, text in documents.items():
        doc_words = {
            normalizer(word) for word in tokenize(text)
        }

        if query_words.issubset(doc_words):
            results.append(doc_id)

    return set(results)


# Evaluation metrics
def evaluate(results):
    tp = len(results & relevant_docs)
    fp = len(results - relevant_docs)
    fn = len(relevant_docs - results)

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return precision, recall, f1


# Compare all three approaches
methods = {
    "Exact Matching": exact_normalize,
    "Porter Stemming": porter_normalize,
    "Morphology-Aware": morphology_normalize
}

print("NORMALISATION COMPARISON")
print("-" * 65)

for method_name, normalizer in methods.items():
    results = search_documents(normalizer)
    precision, recall, f1 = evaluate(results)

    print("\nMethod:", method_name)
    print("Retrieved Documents:", sorted(results))
    print("Precision:", round(precision, 2))
    print("Recall:", round(recall, 2))
    print("F1 Score:", round(f1, 2))


# Final recommendation
print("\nRECOMMENDATION")
print("Use morphology-aware normalisation with")
print("clinical abbreviation and terminology rules.")
print("It improves matching while preserving medical meaning.")
```
