```python
#Experiment 1: Regular Expression Pattern Matching
import re

text = input("Enter the text: ")

# Find first date
date = re.search(r'\b\d{2}[-/]\d{2}[-/]\d{4}\b', text)

if date:
    print("First date:", date.group())
else:
    print("No date found")

# Find mobile numbers
mobiles = re.findall(r'\b[6-9]\d{9}\b', text)
print("Mobile numbers:", mobiles)

# Find email IDs
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)
print("Email IDs:", emails)

# Find capitalized words
capital_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)
print("Capitalized words:", capital_words)

# Check whether text starts with "call"
if re.match(r'(?i)^call\b', text):
    print("Text starts with 'call'")
else:
    print("Text does not start with 'call'")
```

```python
#Experiment-2: Finite State Automaton – String Ending with "ab"
def fsa(string):
    state = "q0"
    path = [state]

    for ch in string:
        if ch not in "ab":
            print("Invalid input! Use only a and b.")
            return

        if state == "q0":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        elif state == "q1":
            if ch == "a":
                state = "q1"
            else:
                state = "q2"

        elif state == "q2":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        path.append(state)

    print("State Path:", " -> ".join(path))

    if state == "q2":
        print("ACCEPTED")
    else:
        print("REJECTED")


string = input("Enter a string over {a,b}: ")
fsa(string)
```

```python
#Experiment-4: Finite State Machine for Plural Forms
def plural(noun):
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        return noun + "es"
    elif noun.endswith("y") and noun[-2] not in "aeiou":
        return noun[:-1] + "ies"
    else:
        return noun + "s"


word = input("Enter an English noun: ")

print("Singular:", word)
print("Plural:", plural(word))
```

```python
#Experiment-6: Bigram Model for Text Generation
from collections import defaultdict
import random

text = input("Enter a sentence: ")

words = text.lower().split()

bigram = defaultdict(list)

for i in range(len(words) - 1):
    bigram[words[i]].append(words[i + 1])

start = words[0]
result = [start]

for i in range(10):
    if result[-1] in bigram:
        next_word = random.choice(bigram[result[-1]])
        result.append(next_word)
    else:
        break

print("Generated text:")
print(" ".join(result))
```

```python
#Program 8: Simple stochastic POS tagger
import nltk
from collections import defaultdict, Counter

training = [
    [("The","DET"),("cat","NOUN"),("runs","VERB")],
    [("A","DET"),("dog","NOUN"),("barks","VERB")],
    [("The","DET"),("boy","NOUN"),("plays","VERB")],
    [("A","DET"),("girl","NOUN"),("sings","VERB")],
    [("cat","NOUN"),("sleeps","VERB")]
]

word_tag_counts = defaultdict(Counter)
tag_counts = Counter()
for sent in training:
    for word, tag in sent:
        word_tag_counts[word.lower()][tag] += 1
        tag_counts[tag] += 1

def stochastic_tag(sentence):
    result = []
    for word in nltk.word_tokenize(sentence):
        counts = word_tag_counts.get(word.lower())
        if counts:
            tag = counts.most_common(1)[0][0]
        else:
            # Simple fallback for words not in training data
            tag = "VERB" if word.lower().endswith(("s", "ed", "ing")) else "NOUN"
        result.append((word, tag))
    return result

sentence = "The dog plays"
print(stochastic_tag(sentence))
```

```python
#Program 9: Rule-based POS tagging using regular expressions
import nltk

patterns = [
    (r".*ing$", "VBG"),       # gerund/present participle
    (r".*ed$", "VBD"),        # past-tense verb
    (r".*ly$", "RB"),         # adverb
    (r".*ous$", "JJ"),        # adjective
    (r".*s$", "NNS"),         # plural noun
    (r".*", "NN")             # default: noun
]

regexp_tagger = nltk.RegexpTagger(patterns)
sentence = "The clever boy is running quickly"
tokens = nltk.word_tokenize(sentence)
print(regexp_tagger.tag(tokens))
```

```python
#Program 10: Transformation-based tagging
# Start with simple initial tags, then apply transformation rules.
words = ["They", "can", "fish", "today"]
initial_tags = ["NN", "NN", "NN", "NN"]
tagged = list(zip(words, initial_tags))

# Rule 1: pronouns are tagged PRP
pronouns = {"i", "you", "he", "she", "it", "we", "they"}
tagged = [(w, "PRP" if w.lower() in pronouns else t) for w, t in tagged]

# Rule 2: common modal verbs are tagged MD
modals = {"can", "could", "may", "might", "must", "shall", "should", "will", "would"}
tagged = [(w, "MD" if w.lower() in modals else t) for w, t in tagged]

# Rule 3: a word after a modal is tagged as a base-form verb
for i in range(1, len(tagged)):
    if tagged[i-1][1] == "MD":
        tagged[i] = (tagged[i][0], "VB")

print("After transformation rules:")
print(tagged)
```

```python
#Program 11: Top-down parser for a context-free grammar
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'boy' | 'dog'
V -> 'sees'
""")

parser = nltk.RecursiveDescentParser(grammar)
sentence = "the boy sees a dog".split()

trees = list(parser.parse(sentence))
for tree in trees:
    print(tree)
    tree.pretty_print()
```

```python
#Program 14: Check subject–verb agreement using CFG rules
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP_SG VP_SG | NP_PL VP_PL
NP_SG -> Det N_SG
NP_PL -> Det N_PL
VP_SG -> V_SG
VP_PL -> V_PL
Det -> 'the' | 'a'
N_SG -> 'boy' | 'girl'
N_PL -> 'boys' | 'girls'
V_SG -> 'runs' | 'plays'
V_PL -> 'run' | 'play'
""")

sentence = "the boy runs".split()
parser = nltk.ChartParser(grammar)
parses = list(parser.parse(sentence))

if parses:
    print("Correct subject–verb agreement.")
    for tree in parses:
        print(tree)
else:
    print("Agreement error or sentence not covered by the grammar.")
```

```python

import re

def tokenize(expression):
    pattern = r'[A-Za-z_][A-Za-z_0-9]*|[(),]'
    return re.findall(pattern, expression)

class FOPCParser:

    def __init__(self, tokens):
        self.tokens = tokens
        self.pos = 0

    def peek(self):
        if self.pos < len(self.tokens):
            return self.tokens[self.pos]
        return None

    def consume(self):
        token = self.peek()
        if token is not None:
            self.pos += 1
        return token

    def parse_atom(self):

        # Handle NOT operator
        if self.peek() == "NOT":
            self.consume()
            return ("NOT", self.parse_atom())

        # Handle parentheses
        if self.peek() == "(":
            self.consume()
            result = self.parse_or()

            if self.consume() != ")":
                raise ValueError("Missing closing parenthesis")

            return result

        # Handle predicate
        name = self.consume()

        if name is None or name in ("AND", "OR", "NOT", ",", ")"):
            raise ValueError("Expected predicate name")

        if self.peek() == "(":
            self.consume()
            args = []

            while self.peek() != ")":
                arg = self.consume()

                if arg is None or arg in ("AND", "OR", "NOT", ",", "("):
                    raise ValueError("Invalid predicate argument")

                args.append(arg)

                if self.peek() == ",":
                    self.consume()
                elif self.peek() != ")":
                    raise ValueError("Expected comma or closing parenthesis")

            self.consume()
            return ("PREDICATE", name, args)

        return ("SYMBOL", name)

    def parse_and(self):
        left = self.parse_atom()

        while self.peek() == "AND":
            self.consume()
            right = self.parse_atom()
            left = ("AND", left, right)

        return left

    def parse_or(self):
        left = self.parse_and()

        while self.peek() == "OR":
            self.consume()
            right = self.parse_and()
            left = ("OR", left, right)

        return left

    def parse(self):
        result = self.parse_or()

        if self.peek() is not None:
            raise ValueError("Unexpected token: " + self.peek())

        return result


# Input expression
expression = "Human(Socrates) AND Mortal(Socrates)"

parser = FOPCParser(tokenize(expression))

print("Expression:", expression)
print("Parsed structure:", parser.parse())
```

```python
#Program 19: Word Sense Disambiguation using the Lesk algorithm
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')

from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentence = "I deposited money in the bank"
ambiguous_word = "bank"

sense = lesk(word_tokenize(sentence), ambiguous_word)
print("Sentence:", sentence)
print("Ambiguous word:", ambiguous_word)
if sense:
    print("Selected sense:", sense.name())
    print("Meaning:", sense.definition())
else:
    print("No sense found.")
```

```python
#Program 20: Information retrieval using TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Python is a popular programming language.",
    "Machine learning uses algorithms to learn from data.",
    "Natural language processing works with human language.",
    "Python is used for data science and machine learning."
]
query = "Python machine learning"

vectorizer = TfidfVectorizer()
doc_vectors = vectorizer.fit_transform(documents + [query])
scores = cosine_similarity(doc_vectors[-1], doc_vectors[:-1]).flatten()

ranking = sorted(enumerate(scores, start=1),
                 key=lambda item: item[1], reverse=True)

print("Documents ranked for query:", query)
for doc_id, score in ranking:
    print(f"Document {doc_id}: score={score:.3f}")
    print(documents[doc_id-1])
```

```python
#Program 22: Simple rule-based reference resolution
import re

text = "Ravi bought a laptop. He uses it every day."
sentences = re.split(r'(?<=[.!?])\\s+', text)
mentions = []

for sent in sentences:
    words = re.findall(r"\\b[A-Za-z]+\\b", sent)
    for word in words:
        if word.lower() not in {"he", "she", "it", "they", "him", "her", "them"}:
            mentions.append(word)

pronouns = {"he": "Ravi", "she": "Ravi", "it": "laptop",
            "him": "Ravi", "her": "Ravi", "they": "previous plural noun",
            "them": "previous plural noun"}

print("Text:", text)
for pronoun in re.findall(r"\\b(he|she|it|they|him|her|them)\\b", text, flags=re.I):
    print(pronoun, "->", pronouns.get(pronoun.lower(), "unknown"))
```

```python
#Program 23: Evaluate text coherence
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = ("Machine learning is a branch of artificial intelligence. "
        "It allows computers to learn patterns from data. "
        "Training data is used to build a predictive model.")

sentences = [s.strip() for s in re.split(r'(?<=[.!?])\\s+', text) if s.strip()]
if len(sentences) < 2:
    print("Enter at least two sentences.")
else:
    matrix = TfidfVectorizer().fit_transform(sentences)
    similarities = cosine_similarity(matrix[:-1], matrix[1:]).diagonal()
    print("Adjacent sentence similarities:")
    for i, score in enumerate(similarities):
        print(f"Sentence {i+1} ↔ {i+2}: {score:.3f}")
    print("Average coherence score:", round(float(similarities.mean()), 3))
    print("Note: This is a simple lexical similarity estimate, not a complete linguistic evaluation.")
```

```python
#Program 24: Dialogue act recognition
def recognize_dialogue_act(sentence):
    s = sentence.lower().strip()
    if any(x in s for x in ["hello", "hi", "good morning", "good evening"]):
        return "GREETING"
    if s.endswith("?") or s.startswith(("what", "where", "when", "why", "who", "how", "can you", "could you")):
        return "QUESTION"
    if any(x in s for x in ["please", "could you", "can you", "kindly"]):
        return "REQUEST"
    if any(x in s for x in ["thank you", "thanks"]):
        return "THANKING"
    if any(x in s for x in ["bye", "goodbye", "see you"]):
        return "FAREWELL"
    return "STATEMENT"

dialogue = [
    "Hello!",
    "What is your name?",
    "Please open the door.",
    "Thank you.",
    "Goodbye!"
]

for utterance in dialogue:
    print(utterance, "->", recognize_dialogue_act(utterance))
```

```python
#Program 26: Machine translation: English to French using Transformers

!pip install -q transformers sentencepiece torch

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load English to French translation model
model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Input English sentence
english_text = "Good morning. How are you?"

# Convert text into tokens
inputs = tokenizer(english_text, return_tensors="pt")

# Generate French translation
translated = model.generate(
    **inputs,
    max_new_tokens=100
)

# Convert tokens into readable text
french_text = tokenizer.decode(
    translated[0],
    skip_special_tokens=True
)

# Display result
print("English Text:", english_text)
print("French Translation:", french_text)
```
