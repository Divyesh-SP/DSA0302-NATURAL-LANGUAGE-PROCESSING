## Program 16: Named Entity Recognition using spaCy

Input: edit the `text` variable.

```python
!pip -q install spacy
!python -m spacy download en_core_web_sm

import spacy
nlp = spacy.load("en_core_web_sm")

text = "Sundar Pichai visited Google in London on Monday."
doc = nlp(text)

print("Named Entities:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```
