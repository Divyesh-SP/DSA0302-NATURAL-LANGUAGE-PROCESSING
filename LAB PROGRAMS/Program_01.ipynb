# Experiment 1: Regular Expressions for Pattern Matching and Searching

**Aim:** To implement regular expressions for pattern matching and searching using Python.

Input: change `text` and `pattern` to search for a pattern in your own text.


```python
import re

text = "Python is easy to learn. I am learning Python programming."
pattern = r"Python"

# Search for the first occurrence anywhere in the text
search_result = re.search(pattern, text)

# Find all occurrences
all_matches = re.findall(pattern, text)

# Match checks only from the beginning of the text
match_result = re.match(pattern, text)

print("Input text:", text)
print("Pattern:", pattern)
print("Search found:", search_result.group() if search_result else "Not found")
print("All matches:", all_matches)
print("Match at beginning:", bool(match_result))
```

**Result:** The program was executed to demonstrate the stated NLP task.
