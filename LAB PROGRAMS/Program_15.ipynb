## Program 15: Probabilistic CFG parsing

Input: change `sentence`; use words listed in the PCFG.

```python
import nltk

pcfg = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.7] | 'John' [0.3]
VP -> V NP [0.6] | V [0.4]
Det -> 'the' [0.6] | 'a' [0.4]
N -> 'dog' [0.5] | 'cat' [0.5]
V -> 'sees' [0.6] | 'runs' [0.4]
""")

sentence = "John sees the dog".split()
parser = nltk.ViterbiParser(pcfg)

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```
