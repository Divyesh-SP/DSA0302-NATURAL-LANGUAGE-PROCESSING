## Program 12: Earley parser for a CFG

Input: change `sentence`; its words must be in the grammar.

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'student' | 'book'
V -> 'reads'
""")

parser = nltk.EarleyChartParser(grammar)
sentence = "the student reads a book".split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```
