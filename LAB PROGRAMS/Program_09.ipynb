## Program 9: Rule-based POS tagging using regular expressions

Input: change the `sentence` variable.

```python
import nltk

patterns = [
    (r".*ing$", "VBG"),       # gerund/present participle
    (r".*ed$", "VBD"),        # past-tense verb
    (r".*ly$", "RB"),         # adverb
    (r".*ous$", "JJ"),        # adjective
    (r".*s$", "NNS"),         # plural noun
    (r".*", "NN")             # default: noun
]

regexp_tagger = nltk.RegexpTagger(patterns)
sentence = "The clever boy is running quickly"
tokens = nltk.word_tokenize(sentence)
print(regexp_tagger.tag(tokens))
```
