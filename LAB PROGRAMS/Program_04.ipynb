# Experiment 4: Finite State Machine for English Noun Plural Formation

**Aim:** To implement finite state machine for english noun plural formation using Python.

Input: change `nouns`. This simple finite-state morphological transducer applies common plural rules: consonant + y → ies; s/x/z/ch/sh → es; otherwise → s.


```python
def plural_fsm(noun):
    word = noun.lower()

    # State q0: inspect the ending and choose a transition
    if len(word) > 1 and word.endswith("y") and word[-2] not in "aeiou":
        state = "q_y"
    elif word.endswith(("s", "x", "z", "ch", "sh")):
        state = "q_sibilant"
    else:
        state = "q_default"

    # Output state: apply the plural-forming rule
    if state == "q_y":
        plural = noun[:-1] + "ies"
    elif state == "q_sibilant":
        plural = noun + "es"
    else:
        plural = noun + "s"

    return state, plural

nouns = ["cat", "bus", "box", "baby", "church", "toy"]

print("Singular\tState\t\tPlural")
for noun in nouns:
    state, plural = plural_fsm(noun)
    print(f"{noun}\t\t{state}\t{plural}")
```
