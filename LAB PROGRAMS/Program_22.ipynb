## Program 22: Simple rule-based reference resolution

Input: change `text`. This is a basic demonstration, not a full coreference-resolution model.

```python
import re

text = "Ravi bought a laptop. He uses it every day."
sentences = re.split(r'(?<=[.!?])\\s+', text)
mentions = []

for sent in sentences:
    words = re.findall(r"\\b[A-Za-z]+\\b", sent)
    for word in words:
        if word.lower() not in {"he", "she", "it", "they", "him", "her", "them"}:
            mentions.append(word)

pronouns = {"he": "Ravi", "she": "Ravi", "it": "laptop",
            "him": "Ravi", "her": "Ravi", "they": "previous plural noun",
            "them": "previous plural noun"}

print("Text:", text)
for pronoun in re.findall(r"\\b(he|she|it|they|him|her|them)\\b", text, flags=re.I):
    print(pronoun, "->", pronouns.get(pronoun.lower(), "unknown"))
```
