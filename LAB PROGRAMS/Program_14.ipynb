## Program 14: Check subject–verb agreement using CFG rules

Input: set `sentence` to a test sentence. Try `the boy runs` and `the boys run`.

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP_SG VP_SG | NP_PL VP_PL
NP_SG -> Det N_SG
NP_PL -> Det N_PL
VP_SG -> V_SG
VP_PL -> V_PL
Det -> 'the' | 'a'
N_SG -> 'boy' | 'girl'
N_PL -> 'boys' | 'girls'
V_SG -> 'runs' | 'plays'
V_PL -> 'run' | 'play'
""")

sentence = "the boy runs".split()
parser = nltk.ChartParser(grammar)
parses = list(parser.parse(sentence))

if parses:
    print("Correct subject–verb agreement.")
    for tree in parses:
        print(tree)
else:
    print("Agreement error or sentence not covered by the grammar.")
```
