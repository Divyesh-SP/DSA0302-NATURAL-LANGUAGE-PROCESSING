## Program 8: Simple stochastic POS tagger

Input: change the `sentence` variable. The small training data is included.

```python
import nltk
from collections import defaultdict, Counter

training = [
    [("The","DET"),("cat","NOUN"),("runs","VERB")],
    [("A","DET"),("dog","NOUN"),("barks","VERB")],
    [("The","DET"),("boy","NOUN"),("plays","VERB")],
    [("A","DET"),("girl","NOUN"),("sings","VERB")],
    [("cat","NOUN"),("sleeps","VERB")]
]

word_tag_counts = defaultdict(Counter)
tag_counts = Counter()
for sent in training:
    for word, tag in sent:
        word_tag_counts[word.lower()][tag] += 1
        tag_counts[tag] += 1

def stochastic_tag(sentence):
    result = []
    for word in nltk.word_tokenize(sentence):
        counts = word_tag_counts.get(word.lower())
        if counts:
            tag = counts.most_common(1)[0][0]
        else:
            # Simple fallback for words not in training data
            tag = "VERB" if word.lower().endswith(("s", "ed", "ing")) else "NOUN"
        result.append((word, tag))
    return result

sentence = "The dog plays"
print(stochastic_tag(sentence))
```
