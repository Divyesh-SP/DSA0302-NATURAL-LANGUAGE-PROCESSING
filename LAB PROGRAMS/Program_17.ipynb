## Program 17: WordNet: synonyms and word meanings

Input: change the `word` variable.

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import wordnet as wn

word = "happy"
synsets = wn.synsets(word)

print("Synsets and meanings:")
for syn in synsets[:5]:
    print("\\nSynset:", syn.name())
    print("Meaning:", syn.definition())
    print("Synonyms:", [lemma.name() for lemma in syn.lemmas()])

print("\\nAll unique synonyms:")
synonyms = sorted({lemma.name().replace("_", " ")
                   for syn in synsets for lemma in syn.lemmas()})
print(synonyms)
```
