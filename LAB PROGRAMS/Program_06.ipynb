# Experiment 6: Bigram N-gram Model for Text Generation

**Aim:** To implement bigram n-gram model for text generation using Python.

Input: change `seed_word` and `number_of_words`. The example corpus is included; add sentences to `corpus` to train on your own text.


```python
import random
from collections import defaultdict

corpus = [
    "natural language processing helps computers understand human language",
    "natural language processing is a part of artificial intelligence",
    "artificial intelligence helps machines learn from data",
    "language models generate text using word patterns"
]

# Build a bigram model: each word stores possible next words
bigram_model = defaultdict(list)
for sentence in corpus:
    words = sentence.lower().split()
    for i in range(len(words) - 1):
        bigram_model[words[i]].append(words[i + 1])

seed_word = "natural"
number_of_words = 12
random.seed(10)  # makes this example repeatable

generated = [seed_word]
current_word = seed_word

for _ in range(number_of_words - 1):
    possible_next_words = bigram_model.get(current_word)
    if not possible_next_words:
        break
    current_word = random.choice(possible_next_words)
    generated.append(current_word)

print("Seed word:", seed_word)
print("Generated text:", " ".join(generated))
```

**Result:** The program was executed to demonstrate the stated NLP task.
