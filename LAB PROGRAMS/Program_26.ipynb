## Program 26: English to French machine translation using Hugging Face Transformers

Input: change `english_text`. Uses Helsinki-NLP/opus-mt-en-fr. The first run downloads the model.

```python
!pip -q install transformers sentencepiece torch

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

english_text = "Good morning. How are you?"
inputs = tokenizer(english_text, return_tensors="pt")
translated = model.generate(**inputs, max_new_tokens=100)
french_text = tokenizer.decode(translated[0], skip_special_tokens=True)

print("English Text:", english_text)
print("French Translation:", french_text)
```
