# Experiment 5: Porter Stemming using NLTK

**Aim:** To implement porter stemming using nltk using Python.

Input: change the `words` list to test other words.


```python
!pip -q install nltk

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["playing", "played", "plays", "connected", "connection", "studies"]

print("Original Word\tStemmed Word")
for word in words:
    print(f"{word}\t\t{stemmer.stem(word)}")
```

**Result:** The program was executed to demonstrate the stated NLP task.
