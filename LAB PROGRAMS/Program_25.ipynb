## Program 25: Free text generation using Hugging Face Transformers (GPT alternative)

Input: change `prompt`. This free alternative uses DistilGPT-2 and does not require an OpenAI API key.

```python
!pip -q install transformers torch

from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural Language Processing is"
result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7,
    pad_token_id=50256
)

print("Input Prompt:", prompt)
print("\nGenerated Text:")
print(result[0]["generated_text"])
```
