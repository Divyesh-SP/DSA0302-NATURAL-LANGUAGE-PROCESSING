## Program 21: Syntax-driven semantic analysis: noun phrases and meanings

Input: change the `text` variable. spaCy model is installed in the cell.

```python
!pip -q install spacy
!python -m spacy download en_core_web_sm
!pip -q install nltk

import spacy
import nltk
from nltk.corpus import wordnet as wn
nltk.download('wordnet')
nltk.download('omw-1.4')

nlp = spacy.load("en_core_web_sm")
text = "The young student reads an interesting book about science."
doc = nlp(text)

print("Noun phrases and semantic information:")
for chunk in doc.noun_chunks:
    head = chunk.root
    synsets = wn.synsets(head.lemma_, pos=wn.NOUN)
    meaning = synsets[0].definition() if synsets else "Meaning not found in WordNet"
    print("\\nNoun phrase:", chunk.text)
    print("Head word:", head.text)
    print("Meaning:", meaning)
```
