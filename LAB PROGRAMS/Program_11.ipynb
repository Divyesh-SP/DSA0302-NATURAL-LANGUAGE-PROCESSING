## Program 11: Top-down parser for a context-free grammar

Input: enter a sentence whose words match the grammar, e.g. `the boy sees a dog`.

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'boy' | 'dog'
V -> 'sees'
""")

parser = nltk.RecursiveDescentParser(grammar)
sentence = "the boy sees a dog".split()

trees = list(parser.parse(sentence))
for tree in trees:
    print(tree)
    tree.pretty_print()
```
