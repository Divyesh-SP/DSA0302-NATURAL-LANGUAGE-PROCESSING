## Program 13: Generate a parse tree using a CFG

Input: change `sentence` to a sentence allowed by the grammar.

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det Adj N | Det N
VP -> V NP
Det -> 'the' | 'a'
Adj -> 'small' | 'big'
N -> 'child' | 'ball'
V -> 'kicks' | 'sees'
""")

sentence = "the small child kicks a ball".split()
parser = nltk.ChartParser(grammar)

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```
