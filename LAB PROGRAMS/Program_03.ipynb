# Experiment 3: Morphological Analysis using NLTK

**Aim:** To implement morphological analysis using nltk using Python.

Input: change `words` to a list of words. This example demonstrates stemming and WordNet lemmatization.


```python
!pip -q install nltk

import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import PorterStemmer, WordNetLemmatizer

words = ["playing", "played", "plays", "studies", "cars", "better"]
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print("Word\tStem\t\tNoun lemma\tVerb lemma")
for word in words:
    stem = stemmer.stem(word)
    noun_lemma = lemmatizer.lemmatize(word, pos="n")
    verb_lemma = lemmatizer.lemmatize(word, pos="v")
    print(f"{word}\t{stem}\t\t{noun_lemma}\t\t{verb_lemma}")
```

**Result:** The program was executed to demonstrate the stated NLP task.
