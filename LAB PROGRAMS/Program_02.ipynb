# Experiment 2: Finite State Automaton for Strings Ending with ab

**Aim:** To implement finite state automaton for strings ending with ab using Python.

Input: change `input_string`. The automaton accepts strings over the alphabet `{a, b}` that end with `ab`.


```python
# DFA states:
# q0 = no useful suffix, q1 = last character is a,
# q2 = string ends with ab (accepting state)

transitions = {
    "q0": {"a": "q1", "b": "q0"},
    "q1": {"a": "q1", "b": "q2"},
    "q2": {"a": "q1", "b": "q0"}
}

start_state = "q0"
final_state = "q2"
input_string = "aabab"

state = start_state
valid_symbols = True

for symbol in input_string:
    if symbol not in ("a", "b"):
        valid_symbols = False
        break
    state = transitions[state][symbol]

if valid_symbols and state == final_state:
    print("Accepted: the string ends with 'ab'.")
else:
    print("Rejected: the string does not end with 'ab' or has invalid symbols.")
```

**Result:** The program was executed to demonstrate the stated NLP task.
