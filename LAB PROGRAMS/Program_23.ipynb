## Program 23: Evaluate text coherence

Input: replace the `text` with your own multi-sentence paragraph.

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = ("Machine learning is a branch of artificial intelligence. "
        "It allows computers to learn patterns from data. "
        "Training data is used to build a predictive model.")

sentences = [s.strip() for s in re.split(r'(?<=[.!?])\\s+', text) if s.strip()]
if len(sentences) < 2:
    print("Enter at least two sentences.")
else:
    matrix = TfidfVectorizer().fit_transform(sentences)
    similarities = cosine_similarity(matrix[:-1], matrix[1:]).diagonal()
    print("Adjacent sentence similarities:")
    for i, score in enumerate(similarities):
        print(f"Sentence {i+1} ↔ {i+2}: {score:.3f}")
    print("Average coherence score:", round(float(similarities.mean()), 3))
    print("Note: This is a simple lexical similarity estimate, not a complete linguistic evaluation.")
```
