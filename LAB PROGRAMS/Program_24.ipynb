## Program 24: Dialogue act recognition

Input: change the `dialogue` list.

```python
def recognize_dialogue_act(sentence):
    s = sentence.lower().strip()
    if any(x in s for x in ["hello", "hi", "good morning", "good evening"]):
        return "GREETING"
    if s.endswith("?") or s.startswith(("what", "where", "when", "why", "who", "how", "can you", "could you")):
        return "QUESTION"
    if any(x in s for x in ["please", "could you", "can you", "kindly"]):
        return "REQUEST"
    if any(x in s for x in ["thank you", "thanks"]):
        return "THANKING"
    if any(x in s for x in ["bye", "goodbye", "see you"]):
        return "FAREWELL"
    return "STATEMENT"

dialogue = [
    "Hello!",
    "What is your name?",
    "Please open the door.",
    "Thank you.",
    "Goodbye!"
]

for utterance in dialogue:
    print(utterance, "->", recognize_dialogue_act(utterance))
```
