## Program 10: Transformation-based tagging

Input: change the `words` list. The transformation rules are shown in the code.

```python
# Start with simple initial tags, then apply transformation rules.
words = ["They", "can", "fish", "today"]
initial_tags = ["NN", "NN", "NN", "NN"]
tagged = list(zip(words, initial_tags))

# Rule 1: pronouns are tagged PRP
pronouns = {"i", "you", "he", "she", "it", "we", "they"}
tagged = [(w, "PRP" if w.lower() in pronouns else t) for w, t in tagged]

# Rule 2: common modal verbs are tagged MD
modals = {"can", "could", "may", "might", "must", "shall", "should", "will", "would"}
tagged = [(w, "MD" if w.lower() in modals else t) for w, t in tagged]

# Rule 3: a word after a modal is tagged as a base-form verb
for i in range(1, len(tagged)):
    if tagged[i-1][1] == "MD":
        tagged[i] = (tagged[i][0], "VB")

print("After transformation rules:")
print(tagged)
```
